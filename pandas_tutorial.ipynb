```python
#AQI Data set
df = pd.read_csv("globalAirQuality.csv")
df.describe()
```

```python
#selecting data

#columns
df['city']
df[['city','aqi']]

#rows
df.loc[2]
df.loc[0:2] #start idx : end idx (inclusive)

df.iloc[2]
df.iloc[0:3] #start idx : end idx (exclusive)

#cells - row,col
df.loc[0:2,"city"]
df.columns

df.iloc[0:3,2]
df.loc[0:2,['country','city','latitude']]
df.iloc[0:3,1:5]
#df.iloc[0:2,['city','aqi']] #it show error because iloc support index values(num) only

#cell
df.at[0,'city']
df.iat[0,2]


#to avoid changes in orignal data we create copy because pandas use view so 
#changes can be made to orignal data (it is recomended to use a copy to avoid changes to orignal data)
cities = df['city'].copy()
cities[0] = 'Mumbai' #copy

cities
```

```python
#filtering of data

df[df['aqi']>100]
df[(df['aqi']>100 & (df['temperature']>30))]

aqi_data = df[(df['aqi']>100) & (df['temperature']>30)][['city','aqi']]
aqi_data
aqi_data.iloc[0]
aqi_data.loc[6]

#Query

df.query('aqi>100')
df.query('aqi>100 & temperature >30') [['city','aqi']]
aqi_val = 100
aqi_data = df.query('aqi>@aqi_val & temperature > 30') [['city','aqi']]
aqi_data
```

```python
#cleaning Data -missing values
df = pd.read_csv('raw_data.csv')
df.isnull()
df.isnull().sum()

df.dropna() #row drop
df.dropna(axis=1) #column drop

df.fillna(0)

age_mean = df['age'].mean()


cleaned_data = df.copy()
age_mean = cleaned_data['age'].mean()
cleaned_data['age'] = cleaned_data['age'].fillna(age_mean)

cleaned_data

df.ffill()
df.bfill()
```

```python
#Data Types
df.dtypes
df2 = df.copy()
df2 = df2.fillna(0)
df2['age'] = df2['age'].astype('int64').copy()
df2.dtypes
```
