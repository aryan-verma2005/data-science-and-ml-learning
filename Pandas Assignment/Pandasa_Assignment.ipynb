```python
import pandas as pd
```

```python
df = pd.read_csv('IRIS.csv')
df.head(10)#Display First 10 rows
df.shape #Display shape
df.dtypes #display datatype
df.describe #Display Summary statistics (mean, std, min, max)

```

```python
df[df['petal_length'] > 4.5] #Select only those rows where: petal_length > 4.5
df[df['species']== 'Iris-virginica']#Select only those rows where:species = "Iris-virginica"
```

```python
# Group by species and compute:-
#Average sepal_length
#Maximum petal_width
#Standard deviation of sepal_width
df.groupby('species').agg(
    avg_sepal_length=('sepal_length', 'mean'),
    max_petal_width=('petal_width', 'max'),
    std_sepal_width=('sepal_width', 'std')
)

 #OR
df.groupby('species').agg(['mean','max','min'])


```

```python
df.groupby('species').agg(['mean','max','min'])
```

```python
#Create a new column:"petal_ratio" = petal_length / petal_width

df['petal_ratio'] = df['petal_length'] / df['petal_width']
df['species']['petal_ratio']

#• For each species, find the average petal_ratio

df.groupby('species')['petal_ratio'].mean()
```
