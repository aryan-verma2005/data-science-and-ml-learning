```python
#Q4
tips=sns.load_dataset('tips')
sns.scatterplot(
    data=tips,
    x='day',
    y='total_bill',
    hue='sex',
    style='smoker',
    
)

plt.title('Bill v/s Days')
plt.xlabel('Dayss')
plt.ylabel('Total Bill Amount')
```

```python
#Q5
pivot_table=tips.pivot_table(
    values="tip",
    index="day",
    columns="time",
    aggfunc="mean"

)

sns.heatmap(
    pivot_table,
    annot=True,
    fmt='.2f'
)
```
